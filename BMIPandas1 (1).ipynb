```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

#%% Importing Data
flights_data = pd.read_csv('flights.csv')
flights_data.head(10)
weather_data_pd = pd.read_csv('weather.csv')
weather_data_np = weather_data_pd.to_numpy()
#%% Pandas Data Filtering/Sorting Question Answering
#use flights_data

#Question 1 How many flights were there from JFK to SLC? Int
q_1 = ((flights_data['origin'] == 'JFK') & (flights_data['dest'] == 'SLC')).sum()
print(f"P1: {q_1}")
#Question 2 How many airlines fly to SLC? Should be int
toSLC = flights_data[flights_data['dest'] == 'SLC']
q_2 = toSLC['carrier'].nunique()
print(f"P2: {q_2}")
#Question 3 What is the average arrival delay for flights to RDU? float
toRDU = flights_data[flights_data['dest'] == 'RDU']
q_3 = toRDU['arr_delay'].mean()
print(f"P3: {q_3}")
#Question 4 What proportion of flights to SEA come from the two NYC airports (LGA and JFK)?  float
toSEA = flights_data[flights_data['dest'] == 'SEA']
sumLGAJFK = ((toSEA['origin'] == 'LGA') | (toSEA['origin'] == 'JFK')).sum()
sumSEATotal = toSEA.shape[0]
q_4 = sumLGAJFK / sumSEATotal
print(f"P4: {q_4}")
#Question 5 Which date has the largest average depature delay? Pd slice with date and float
#please make date a column. Preferred format is 2013/1/1 (y/m/d)
flights_data['date'] = '0'
for index, row in flights_data.iterrows():
    flights_data.at[index, 'date'] = str(row['year']) + "/" + str(row['month']) +  "/" + str(row['day'])
highestMean = 0
highestMeanDate = '0'
uniqueDatesList = flights_data['date'].unique()
for Date in uniqueDatesList:
    tempDf = flights_data.loc[flights_data['date'] == Date]
    tempMean = tempDf['dep_delay'].mean()
    if tempMean > highestMean:
        highestMean = tempMean
        highestMeanDate = Date
q_5 = highestMeanDate
print(f"P5: {q_5}")
#Question 6 Which date has the largest average arrival delay? pd slice with date and float
highestMean = 0
highestMeanDate = '0'
uniqueDatesList = flights_data['date'].unique()
for Date in uniqueDatesList:
    tempDf = flights_data.loc[flights_data['date'] == Date]
    tempMean = tempDf['arr_delay'].mean()
    if tempMean > highestMean:
        highestMean = tempMean
        highestMeanDate = Date
q_6 = highestMeanDate
print(f"P6: {q_6}")
#Question 7 Which flight departing LGA or JFK in 2013 flew the fastest? pd slice with tailnumber and speed
#speed = distance/airtime
tempDf = flights_data.loc[(flights_data['origin'] == 'LGA') | (flights_data['origin'] == 'JFK')]
maxSpeed = 0
maxSpeedNumber = '0'
for index, row in tempDf.iterrows():
    if row['distance'] / row['air_time'] > maxSpeed:
        maxSpeed = row['distance'] / row['air_time']
        maxSpeedNumber = row['tailnum']
q_7 = maxSpeedNumber
print(f"P7: {q_7}")
#Question 8 Replace all nans in the weather pd dataframe with 0s. Pd with no nans
weather_data_pd = weather_data_pd.fillna(0)
q_8 = weather_data_pd
print(f"P8: {q_8.head()}")
#%% Numpy Data Filtering/Sorting Question Answering
#Use weather_data_np
#Question 9 How many observations were made in Feburary? Int
tempDf = weather_data_pd.loc[weather_data_pd['month'] == 2.0]
q_9 = tempDf.shape[0]
print(f"P9: {q_9}")
#Question 10 What was the mean for humidity in February? Float
q_10 = tempDf['humid'].mean()
print(f"P10: {q_10}")
#Question 11 What was the std for humidity in February? Float
q_11 = tempDf['humid'].std()
print(f"P11: {q_11}")
```
